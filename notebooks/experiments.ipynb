```python
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
if not os.getenv("GROQ_API_KEY"):
    raise ValueError("Set GROQ_API_KEY in your environment/.env before running this notebook.")
```

```python
if not os.getenv("GOOGLE_API_KEY"):
    raise ValueError("Set GOOGLE_API_KEY in your environment/.env before running this notebook.")
```

```python
mem0_config = {
    "llm": {
        "provider": "groq",
        "config": {
            "model": "qwen/qwen3.8-27b",
            "api_key": os.environ["GROQ_API_KEY"],
            "temperature": 0,
        },
    },
    "embedder": {
        "provider": "gemini",
        "config": {
            "model": "models/gemini-embedding-001",
            "embedding_dims": 1536,
            "api_key": os.environ["GOOGLE_API_KEY"],
        },
    },
    "vector_store": {
        "provider": "qdrant",
        "config": {
            "path": ":memory:",
            "embedding_model_dims": 1536,
        },
    }
}
```

```python
from mem0 import Memory

m = Memory.from_config(mem0_config)
```

```python
messages = [
    {"role": "user", "content": "Hi, I'm Alex. I love basketball and gaming."},
    {"role": "assistant", "content": "Hey Alex! I'll remember your interests."}
]
m.add(messages, user_id="alex")
```

```python
results = m.search("What do you know about me?", user_id= "alex")
print(results)
```
